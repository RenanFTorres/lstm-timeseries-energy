```python
Para extrair páginas de um arquivo PDF com base em uma lista de códigos usando Python, você pode utilizar a biblioteca PyPDF2. Aqui está um exemplo de código que realiza essa tarefa:


import os
from PyPDF2 import PdfReader, PdfWriter

def extrair_paginas_pdf(arquivo_pdf, codigos_paginas, nome_saida):
    # Abrir o arquivo PDF
    pdf = PdfReader(arquivo_pdf)
    
    # Verificar se os códigos de página estão dentro do intervalo válido
    num_paginas = len(pdf.pages)
    for codigo in codigos_paginas:
        if codigo < 1 or codigo > num_paginas:
            raise ValueError(f"Código de página inválido: {codigo}. O arquivo PDF tem {num_paginas} páginas.")
    
    # Extrair as páginas especificadas
    pdf_saida = PdfWriter()
    for codigo in codigos_paginas:
        pdf_saida.addPage(pdf.pages[codigo - 1])
    
    # Salvar o novo arquivo PDF
    with open(nome_saida, "wb") as saida_pdf:
        pdf_saida.write(saida_pdf)

# Parâmetros
arquivo_pdf = "arquivo_original.pdf"  # Nome do arquivo PDF original
codigos_paginas = [1, 3, 5]  # Lista de códigos de página para extrair
nome_saida = "arquivo_extraido.pdf"  # Nome do arquivo PDF de saída

# Executar a função
extrair_paginas_pdf(arquivo_pdf, codigos_paginas, nome_saida)


Instruções:

1. Instale a biblioteca PyPDF2 executando pip install PyPDF2 no terminal.
2. Substitua "arquivo_original.pdf" pelo caminho para o seu arquivo PDF.
3. Altere a lista codigos_paginas para incluir os códigos das páginas que deseja extrair. Os códigos são baseados em 1, ou seja, a primeira página é 1, a segunda é 2, e assim por diante.
4. Execute o script.

O arquivo PDF resultante será salvo na mesma pasta que o script Python, com o nome especificado em nome_saida. Se desejar salvar em outra pasta, forneça o caminho completo em nome_saida.

Lembre-se de que o código de página começa em 1. Portanto, se você quiser extrair a primeira, terceira e quinta páginas, use [1, 3, 5].
```
