```python
import os
import re
from PyPDF2 import PdfReader, PdfWriter
```

```python
# Caminho do arquivo de texto e do pdf
caminho_txt = './PDF/unidades_cgjp.txt'

arquivo_pdf = './PDF/Completo.pdf'
```

```python
numbers = []
x = 0
# função de ler o TXT e transformar em lista na variável numbers
def read_txt_file(file_path):
    numbers = []

    with open(file_path, 'r') as file:
        for line in file:
            numbers.append(line.strip())

    return numbers
```

```python
# Função para buscar strings em um arquivo PDF e extrair as páginas onde elas estão
def extrair_paginas_por_string(arquivo_pdf, strings_busca, pasta_saida, nome_saida):
    # Carregar o arquivo PDF
    pdf = PdfReader(arquivo_pdf)
    num_paginas = len(pdf.pages)
    
    # Lista para armazenar os números das páginas que contêm as strings buscadas
    paginas_encontradas = set()
    
    # Iterar por cada página e buscar as strings
    for numero_pagina in range(num_paginas):
        pagina_atual = pdf.pages[numero_pagina]
        texto_pagina = pagina_atual.extract_text()
        
        # Verificar se alguma das strings está presente na página
        for string in strings_busca:
            if re.search(string, texto_pagina, re.IGNORECASE):  # Busca case-insensitive
                paginas_encontradas.add(numero_pagina + 1)  # +1 porque a contagem é baseada em 1
    
    if not paginas_encontradas:
        raise ValueError("Nenhuma das strings foi encontrada no PDF.")
    
    # Criar um novo arquivo PDF com as páginas encontradas
    pdf_saida = PdfWriter()
    for pagina in sorted(paginas_encontradas):
        pdf_saida.add_page(pdf.pages[pagina - 1])

    # Garantir que a pasta de saída exista
    if not os.path.exists(pasta_saida):
        os.makedirs(pasta_saida)

    # Combinar o caminho da pasta com o nome do arquivo de saída
    caminho_completo_saida = os.path.join(pasta_saida, nome_saida)
    
    # Salvar o novo arquivo PDF no caminho completo
    with open(caminho_completo_saida, "wb") as saida_pdf:
        pdf_saida.write(saida_pdf)

    print(f"PDF gerado com sucesso: {nome_saida}. Páginas extraídas: {sorted(paginas_encontradas)}")
```

```python
# Criação da lista a partir do TXT
numbers = read_txt_file(caminho_txt)

# Nome do arquivo PDF de saída
nome_saida = "arquivo_extraido.pdf" 

pasta_saida = './PDF/'

# Executar a função
extrair_paginas_por_string(arquivo_pdf, numbers, pasta_saida, nome_saida)
```
