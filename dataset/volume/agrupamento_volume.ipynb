```python
import pandas as pd
import os

def processar_dados_volume(arquivo_caminho):
    """
    Processa os dados de volume e pluviometria de um arquivo xlsx para agrupamento a cada 3 dias.
    Salva o arquivo processado no formato CSV no mesmo diretório com o nome 'volume_grouped_3_days.csv'.

    Parâmetros:
        arquivo_caminho (str): Caminho completo para o arquivo .xlsx a ser processado.
    """
    try:
        # Carregar o arquivo Excel
        dados = pd.read_excel(arquivo_caminho)
        
        # Garantir que as colunas esperadas estão no arquivo
        if not {'DATA', 'COTA', 'PLUV'}.issubset(dados.columns):
            raise ValueError("O arquivo deve conter as colunas: 'DATA', 'COTA' e 'PLUV'")
        
        # Renomear e transformar a coluna de data
        dados.rename(columns={'DATA': 'timestamp'}, inplace=True)
        dados['timestamp'] = pd.to_datetime(dados['timestamp'], format='%d/%m/%Y')
        
        # Padronizar as colunas COTA e PLUV para 2 casas decimais
        dados['COTA'] = dados['COTA'].round(2)
        dados['PLUV'] = dados['PLUV'].round(2)
        
        # Definir o índice como timestamp
        dados.set_index('timestamp', inplace=True)
        
        # Agrupamento a cada 3 dias (média aritmética)
        dados_grouped = dados.resample('3D').mean()
        
        # Resetar o índice para salvar o timestamp como coluna
        dados_grouped.reset_index(inplace=True)
        
        # Determinar o caminho de saída no mesmo diretório
        diretorio, _ = os.path.split(arquivo_caminho)
        caminho_saida = os.path.join(diretorio, 'volume_grouped_3_days.csv')
        
        # Salvar o arquivo no formato CSV
        dados_grouped.to_csv(caminho_saida, index=False, float_format="%.2f")
        
        print(f"Arquivo processado salvo em: {caminho_saida}")
    
    except Exception as e:
        print(f"Ocorreu um erro ao processar o arquivo: {e}")

```

```python
# Exemplo de uso:
processar_dados_volume(r"C:\Users\Renan Torres\OneDrive\Documentos\Github\dataset\volume\Volume - duas unas - 10meses.xlsx")
```
